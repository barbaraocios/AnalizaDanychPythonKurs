Funkcja `walk` z modułu os, pozwala na rekurencyjne wylistowanie zawartości katalogu. Funkcja ta zwraca generator, zwracający trójki postaci:
(ścieżka, lista katalogów zawartych w danej ścieżce, lista plików zawartych w danej ścieżce).

Wykorzystajmy funkcjonalność f-string do czytelnego wypisania zawartości katalogu:

```python
for path, subdirs, files in os.walk(DATA_DIR / "lektury"):
    print(f"{path} contains\n\tsubdirs:{subdirs}\n\tand files:{files}\n")
```

Z funkcji `open` możemy korzystać zarówno jako klasycznej funkcji, jak i menedżera kontekstu. Ta druga metoda ceduja na Pythona zadanie zamknięcia raz otwartego pliku.

Spróbujmy odczytać treść pliku, i zerknąć na jego początek:

```python
filepath = DATA_DIR / "lektury" / "dramaty" / "ajschylos" / "oresteja.txt"

handle = open(filepath)
txt = handle.read()
handle.close()

with open(filepath) as handle:
    txt2 = handle.read()

assert txt == txt2
print(txt[:1000])
```

Możemy wykorzystać elementy tekstu, aby odnaleźć początek faktycznej treści utworu:

```python
first_line = "/ Ściana tylna sceny przedstawia zamek Atrydów w Argos. /"
text_start = txt.index(first_line)
print(text_start)
```

```python
print(txt[-2000:])
```

```python
last_line = "Wy nućcie pean błogi!"
txt.count(last_line)
```

Chcemy postąpić analogicznie w wypadku końca tekstu, jednak ostatnia fraza pojawia się w nim dwa razy. Musimy odnaleźc indeks ostatniego wystąpienia:

```python
first_occ = txt.index(last_line)
line_len = len(last_line)
after_first = txt[first_occ + line_len:]
second_occ = first_occ + line_len + after_first.index(last_line)
text_end = second_occ+line_len
print(txt[text_end - 1000:text_end])
```

Tekst ma zgrubsza uporządkowaną strukturę:

1. Wypowiedzi przedzielone są trzema znakami nowej linii: `\n`
1. Każda wypowiedź jest poprzedzona nazwą mówcy, po której następują dwa znaki nowej linii
1. Strofy wypowiedzi są przedzielone podwójnymi znakami nowej linii

Wpierw odetnijmy nagłówki i stopkę, i usuńmy okalające białe znaki metodą `strip`:

```python
core_text = txt[text_start:text_end].strip()
```

Następnie wykorzystajmy zauważone prawidłowości do podziału tekstu na mowy.

W wypadku gdy nasze spostrzeżenia nie zostaną spełnione przez fragment tekstu, pomijamy go:

```python
data = []
speeches = core_text.split("\n\n\n")
for speech in speeches:
    if speech:
        try:
            speaker, content = speech.split("\n\n", 1) # we make only 1 cut
        except ValueError:
            # fragment does not match our assumptions, we skip it
            continue
        chunks = content.split("\n\n")
        data.append((speaker, chunks))

print(f"{len(data)}/{len(speeches)}") # number of speeches extracted succesfully
print(data[100])
```
